## 1. Variables

text is a `str`, a whole number is an `int`, and `True`/`False` is a `bool`.
Here we store facts about our two datasets and the minimum size the project brief requires.

## 2. Lists

## 3. Dictionaries

A **dictionary** stores **key: value** pairs inside curly brackets `{ }`. You look a value up by its key instead of by position.
We use these dictionaries:

1. `reported_responses`: survey year mapped to the number of responses the source reports.
2. `question_labels`: each survey question code (the key) mapped to what it measures (the value).
3. `objectives`: each of our 3 specific objectives, with its details and the question codes it needs. (A dictionary can hold other dictionaries and lists.)
4. `research_question_links`: each research question number mapped to the objective it serves.
5. `items_per_question` and `q13_columns`: how our question codes map to the **real column names** in the data (explained below).

*Note: the question codes and objective details are copied from our proposal. In the raw files, several codes are really groups of items. For example `Q18` is the twelve columns `Q18a` to `Q18l`. Question 13 (which AI chatbots a student used) is stored differently in the two waves: one column `Q13` in 2023-24, but six columns `Q13a` to `Q13f` in 2024-25.*

```python
# 1) Survey year -> responses reported by the source (built from the variables in Section 1)
reported_responses = {"2023-24": responses_2023_24, "2024-25": responses_2024_25}

# 2) Question code -> what it measures (from the proposal)
question_labels = {
    "Q10":  "Field of study",
    "Q13":  "AI chatbots used / ChatGPT use",
    "Q14":  "Free or paid version of ChatGPT",
    "Q15":  "Overall ChatGPT use",
    "Q18":  "ChatGPT use by study task",
    "Q20":  "Awareness of institutional ChatGPT policy",
    "Q22":  "Ethical concerns about ChatGPT",
    "Q24":  "Satisfaction with ChatGPT",
    "Q26":  "Perceived academic benefits",
    "Q27":  "Perceived academic benefits",
    "Q32":  "Emotions while using ChatGPT",
    "Q33":  "Perceived study difficulty",
    "Q35a": "Self-rated study success",
}

# Objective 3 also needs the survey year, which is a variable WE will add, so we add a new key
question_labels["year"] = "Survey year (wave)"

# 3) Objectives: a dictionary of dictionaries
objectives = {
    1: {
        "short_title": "Perceived academic benefits",
        "description": ("Compare perceived academic benefits across levels of ChatGPT use, "
                        "subscription types and study tasks"),
        "serves": "Students",
        "variables": ["Q13", "Q14", "Q15", "Q18", "Q24", "Q26", "Q27", "Q35a"],
    },
    2: {
        "short_title": "Emotional experiences",
        "description": ("Compare emotional experiences across levels of ChatGPT use "
                        "and perceived study difficulty"),
        "serves": "Students",
        "variables": ["Q15", "Q32", "Q33"],
    },
    3: {
        "short_title": "Use, concerns and policy awareness",
        "description": ("Compare use of AI chatbots, concerns about ChatGPT and awareness of "
                        "ChatGPT policies across survey years and study backgrounds"),
        "serves": "Administrators",
        "variables": ["Q13", "Q15", "Q18", "Q20", "Q22", "Q10", "year"],
    },
}

# 4) Research question number -> objective it serves (from the proposal's table)
research_question_links = {1: 3, 2: 1, 3: 1, 4: 2, 5: 3, 6: 1, 7: 3}

# 5) Real column names in the raw files
#    a) codes that are a group of lettered items: how many items each has (Q18 -> Q18a ... Q18l)
items_per_question = {"Q18": 12, "Q22": 10, "Q24": 7, "Q26": 10, "Q27": 10, "Q32": 15}
#    b) codes that are one single column
single_columns = ["Q10", "Q14", "Q15", "Q20", "Q33", "Q35a"]
#    c) Q13 is stored differently in each wave
q13_columns = {
    "2023-24": ["Q13"],
    "2024-25": ["Q13a", "Q13b", "Q13c", "Q13d", "Q13e", "Q13f"],
}

# Looking things up by key
print("Q24 measures         :", question_labels["Q24"])
print("Objective 2 serves   :", objectives[2]["serves"])
print("Objective 1 variables:", objectives[1]["variables"])
print("2024-25 Q13 columns  :", q13_columns["2024-25"])

# .get() is a safe lookup: it returns a fallback instead of crashing when the key is missing
print("Q99 measures         :", question_labels.get("Q99", "not in our dictionary"))
print("Codes we track       :", len(question_labels))
```

## 4. Loops

A **`for` loop** repeats a block of code once for every item in a list or dictionary.
The indented lines belong to the loop. Loops can sit inside loops (a *nested* loop).

```python
# Loop 1 (nested): print every objective and, inside it, every variable it needs
for number in objectives:                      # goes through the keys 1, 2, 3
    details = objectives[number]
    print(f"Objective {number}: {details['short_title']} (serves {details['serves']})")
    for code_name in details["variables"]:     # inner loop over that objective's list
        print(f"    {code_name:5} -> {question_labels[code_name]}")
    print()
```

```python
# Loop 2: build a list of the UNIQUE variables across all objectives.
# Some codes (like Q15) appear in more than one objective, so we only add a code if it is not already there.
all_variables = []
for number in objectives:
    for code_name in objectives[number]["variables"]:
        if code_name not in all_variables:
            all_variables.append(code_name)

print("Unique variables our objectives need:", len(all_variables))
print(all_variables)

# Loop 3: count how many research questions serve each objective
questions_per_objective = {1: 0, 2: 0, 3: 0}
for rq_number, objective_number in research_question_links.items():
    questions_per_objective[objective_number] += 1    # += 1 adds one to the running count

print()
print("Research questions per objective:", questions_per_objective)
```

```python
# Loop 4: turn our question codes into REAL column names.
# Single codes map to themselves; a grouped code like Q18 becomes Q18a, Q18b, ... using the first n letters.
letters = "abcdefghijklmno"
question_columns = {}

for code_name in single_columns:
    question_columns[code_name] = [code_name]

for code_name, n_items in items_per_question.items():
    columns = []
    for letter in letters[:n_items]:           # letters[:12] = the first 12 letters, a to l
        columns.append(code_name + letter)     # "Q18" + "a" -> "Q18a"
    question_columns[code_name] = columns

print("Q18 expands to:", question_columns["Q18"])
print("Q35a expands to:", question_columns["Q35a"])
```

```python
# Loop 5: for each survey year, list every real column our objectives need.
# Q13 depends on the year, and 'year' is a column we create ourselves, so they get special handling.
needed_columns = {}
for year in survey_years:
    names = []
    for code_name in all_variables:
        if code_name == "year":
            continue                                  # 'continue' = skip to the next code (not in the raw files)
        elif code_name == "Q13":
            names.extend(q13_columns[year])           # .extend() adds every item of a list
        else:
            names.extend(question_columns[code_name])
    needed_columns[year] = names
    print(f"{year}: our objectives need {len(names)} real columns")
```

## 5. Conditionals

A **conditional** (`if` / `elif` / `else`) lets the code choose what to do depending on whether a condition is `True`.
Python runs the first block whose condition is true. We use conditionals to check our plan against the requirements of the project brief.

```python
# Conditional 1 (inside a loop): does each wave's reported size meet the brief's minimum number of rows?
for year, n_responses in reported_responses.items():
    if n_responses >= min_rows_required:
        print(f"OK: {year} reports {n_responses:,} responses, at least {min_rows_required:,}.")
    else:
        print(f"PROBLEM: {year} reports only {n_responses:,} responses, we need {min_rows_required:,}.")

# Conditional 2 (if / elif / else): do we have enough research questions?
# The brief asks for at least 5 answerable research questions.
total_questions = len(research_question_links)
if total_questions > 5:
    print(f"OK: we have {total_questions} research questions (the brief needs at least 5), with spare ones.")
elif total_questions == 5:
    print("OK: we have exactly the 5 research questions the brief needs.")
else:
    print(f"PROBLEM: only {total_questions} research questions, we need at least 5.")
```

## 6. Functions

A **function** is a reusable block of code with a name. We **define** it once with `def`, then **call** it as often as we like.
It takes *inputs* (parameters), does something, and can `return` a result.

We write two small ones that we will reuse later in the project:

* `check_dataset_size()`: checks a rows/columns count against the brief's minimum (in Section 7 we feed it the real `df.shape`).
* `summarise_objective()`: turns one entry of the `objectives` dictionary into a readable sentence (useful for the README and report).

```python
def check_dataset_size(n_rows, n_columns, min_rows=min_rows_required, min_columns=min_columns_required):
    """Print PASS or FAIL and return True/False: does the dataset meet the brief's minimum size?"""
    problems = []
    if n_rows < min_rows:
        problems.append(f"only {n_rows:,} rows (need {min_rows:,})")
    if n_columns < min_columns:
        problems.append(f"only {n_columns} columns (need {min_columns})")

    if problems:
        print("FAIL: " + " and ".join(problems))
        return False
    print(f"PASS: {n_rows:,} rows and {n_columns} columns (need {min_rows:,} x {min_columns}).")
    return True


def summarise_objective(number):
    """Return a one-sentence summary of one project objective, built from the objectives dictionary."""
    if number not in objectives:
        return f"Objective {number} does not exist. Choose from {list(objectives.keys())}."
    details = objectives[number]
    variable_list = ", ".join(details["variables"])
    return (f"Objective {number} (serves {details['serves']}): {details['description']}. "
            f"It uses {len(details['variables'])} variables: {variable_list}.")
```

```python
# --- Using check_dataset_size ---
# Planned check: the rows each source reports, and the real columns our objectives need.
# (Section 7 repeats the check on the loaded data using df.shape.)
for year in survey_years:
    print(year, end=": ")
    check_dataset_size(reported_responses[year], len(needed_columns[year]))

# A deliberately too-small, made-up example, just to show the FAIL path works
print("Made-up test: ", end="")
check_dataset_size(9000, 8)

# --- Using summarise_objective inside a loop ---
print()
for number in objectives:
    print(summarise_objective(number))
    print()

# An objective that does not exist
print(summarise_objective(4))
```

## 7. Loading the data and proving its size (Week 1, item 5)

Now we load the two raw files with Pandas. Things to know:

* `import pandas as pd` loads the Pandas library under the short name `pd`.
* `pd.read_csv(path)` reads a CSV file into a **DataFrame** (a table). We call it `df`.
* The path is **relative**: `../data/raw/` means "go up one folder from `notebooks/`, then into `data/raw/`". It works on any computer that has our repository layout.
* `low_memory=False` makes Pandas read the whole file at once, so each column's type is guessed consistently.
* `df.shape` gives `(rows, columns)`, `df.columns` lists the column names and `df.head()` shows the first five rows.

*The two files must be saved as `data/raw/raw_2023-2024.csv` and `data/raw/raw_2024-2025.csv`.*

```python
import pandas as pd

raw_folder = "../data/raw/"
file_names = {"2023-24": "raw_2023-2024.csv", "2024-25": "raw_2024-2025.csv"}

dataframes = {}                       # a dictionary: survey year -> DataFrame
for year in survey_years:
    dataframes[year] = pd.read_csv(raw_folder + file_names[year], low_memory=False)
    print(f"Loaded {year} from {raw_folder}{file_names[year]}")
```

```python
# Proof of size, column names and a first look, for each wave
for year, df in dataframes.items():
    print("=" * 70)
    print(f"Survey year {year}")
    print("=" * 70)
    print("Shape (rows, columns):", df.shape)
    print()
    print("Columns:")
    print(list(df.columns))
    print()
    print("First five rows:")
    print(df.head())
    print()
    check_dataset_size(df.shape[0], df.shape[1])
    print()
```

### Do the two files line up?

Two files can only be treated as one dataset if their columns correspond. A **set** is a collection of unique items, and Python can compare two sets directly:
`&` gives the items in both, and `-` gives the items in the first but not the second.

```python
columns_2023_24 = set(dataframes["2023-24"].columns)
columns_2024_25 = set(dataframes["2024-25"].columns)

print("Columns in both files :", len(columns_2023_24 & columns_2024_25))
print("Only in 2023-24       :", sorted(columns_2023_24 - columns_2024_25))
print("Only in 2024-25       :", sorted(columns_2024_25 - columns_2023_24))
```

```python
# Do the real columns our objectives need (Loop 5) actually exist in each file?
for year, df in dataframes.items():
    missing = []
    for name in needed_columns[year]:
        if name not in df.columns:
            missing.append(name)

    if len(missing) == 0:
        print(f"{year}: all {len(needed_columns[year])} columns we need are present.")
    else:
        print(f"{year}: MISSING columns {missing}")
```

```python
# Combine the two waves into one analysis table, adding the 'year' column we planned.
# pd.concat() stacks tables on top of each other. ignore_index=True renumbers the rows 0, 1, 2, ...
# Note: this keeps ALL columns from both files, so Q13 (2023-24) and Q13a-Q13f (2024-25) both appear with blanks
# in the other wave. We will line these up properly in the cleaning step.
frames = []
for year, df in dataframes.items():
    copy = df.copy()              # work on a copy so the original tables stay untouched
    copy["year"] = year           # new column: which wave each row comes from
    frames.append(copy)

combined = pd.concat(frames, ignore_index=True)

print("Combined shape (rows, columns):", combined.shape)
print(combined["year"].value_counts())
print()
check_dataset_size(combined.shape[0], combined.shape[1])
```
