#### Note: Reading `describe(include="all")`: where it is appropriate

`describe` returns two kinds of summary. Text columns (`Q1`, `Q4`, `Q5`) get `count`, `unique`, `top` and `freq`. Number columns get `mean`, `std`, the quartiles, `min` and `max`. Most of those number columns are not real quantities, though. They are answer codes.

- **Nominal codes** (e.g. `Q2` gender, `Q10` field of study): the mean and standard deviation are meaningless. A mean gender of 1.62 describes no one. Use `value_counts()` instead.
- **Ordinal codes** (the 1–5 agreement and frequency scales): the median and quartiles are defensible, because the codes are ranked. The mean is only a rough guide, because the gaps between categories are not guaranteed to be equal.
- **`Q3` (age)** is the only true numeric column. Its median (21) and quartiles (19–23) look sensible, but the mean (27.4) and standard deviation (91.5) are inflated by invalid entries: the minimum is 2 and the maximum is 2006, which is a birth year rather than an age. This is a data-quality issue to record, not to fix here.
- **`Q5` (institution)** is free text with 8,725 distinct values, so the same university is likely spelt in several ways.

The same cautions apply as for 2023–24: means and standard deviations are meaningless for nominal codes and only a rough guide for the 1–5 scales.

The difference this year is **`Q3` (age)**. It is stored as text (`str`), because two answers were typed with words (`"20Years"`, and `"19"` followed by the Arabic word for "year"). So `describe` treats it as a category: 64 distinct values, the most common being 20 (3,924 students), and no mean or quartiles. It needs converting to numbers (e.g. `pd.to_numeric(..., errors="coerce")`) during Week 3 cleaning before it can be summarised. We record that here rather than changing the source data.
